```python
import pandas as pd
```

```python
df = pd.read_csv('train.csv')
```

```python
bins = [0, 12, 18, 60, 100]
labels = ['Child', 'Teen', 'Adult', 'Senior']
```

```python
df['AgeGroup'] = pd.cut(df['Age'], bins=bins, labels=labels, right=False)
```

```python
df
```

```python
df = pd.get_dummies(df, columns=['AgeGroup'], drop_first=True)
```
