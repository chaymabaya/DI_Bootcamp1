Exercice 4 : Visualisation des relations par paires avec Seaborn Pair Plot

```python
df.replace({"Choose your gender" : {'Female' : 1 , "Male" : 0} } , inplace= True)
```

```python
df['Your current year of Study'].unique()
```

```python
df['Your current year of Study'] = df['Your current year of Study'].str.lower() 
```

```python
df['Your current year of Study'].unique()
```

```python
df.replace({"Your current year of Study" : {"year 1" : 1 , "year 2" : 2 , "year 3" : 3 , "year 4" : 4 }} , inplace= True)
```

```python
df.replace({"Do you have Depression?" : {'Yes' : 1 , "No" : 0} } , inplace= True)
df.replace({"Do you have Anxiety?" : {'Yes' : 1 , "No" : 0} } , inplace= True)
df.replace({"Do you have Panic attack?" : {'Yes' : 1 , "No" : 0} } , inplace= True)
```

```python
df
```

```python
df_numeric= df.select_dtypes(include= ["int64","float64" ])
```

```python
df_numeric.columns
```

```python
sns.pairplot(df_numeric , hue = "Do you have Depression?")
```

Exercice 5 : Création d’une carte thermique pour visualiser les corrélations

```python
df_corr = df[[
    "Age",
    "Your current year of Study",
    "Do you have Depression?",
    "Do you have Anxiety?",
    "Do you have Panic attack?"
]]
```

```python
corr_matrix = df_corr.corr()
corr_matrix
```

```python
sns.heatmap(
    corr_matrix,
    annot=True,        
    cmap="coolwarm",   
    linewidths=0.5,    
    fmt=".2f"          
)
```

```python
df["CGPA_num"] = df["What is your CGPA?"].apply(
    lambda x: (float(x.split("-")[0]) + float(x.split("-")[1])) / 2
)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
g = sns.FacetGrid(df, col="Do you have Depression?", height=4, aspect=1.2)
g.map_dataframe(sns.histplot, x="CGPA_num", kde=True, bins=20)

g.set_titles(col_template="Dépression : {col_name}")
g.set_axis_labels("CGPA", "Fréquence")


plt.show()
```
