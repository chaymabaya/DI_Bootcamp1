Exercice 2 : Creating a Line Plot for Temperature Variation

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
temperatures = [72, 74, 76, 80, 82, 78, 75]
days = ["Lundi", "Mardi", "Mercredi", "Jeudi", "vendredi", "Samedi", "Dimanche"]
```

```python
plt.figure(figsize=(10,5))
plt.plot(days , temperatures , marker='+' )
plt.xlabel ('Days')
plt.ylabel('temperatures')
plt.title("Temperature Variation Over a Week")
```

Exercice 3 : Visualizing Monthly Sales with a Bar Chart


```python
sales = ([5000, 5500, 6200, 7000, 7500])
months = ["Jan", "Feb", "Mar", "Apr", "May"]
plt.figure(figsize=(10, 5))
plt.bar(months, sales)
plt.xlabel("Mouths")
plt.ylabel("Sales Amount ($)")
plt.title("Monthly Sales for Retail Store")
plt.show()
```

```python
df = pd.read_csv("Student Mental health.csv")
```

```python
df
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df, x='What is your CGPA?', color='Red', discrete=True, shrink=0.8)
plt.title("Distribution des moyennes cumulatives des étudiants", fontsize=14)
plt.xlabel("Plages de CGPA")
plt.ylabel("Nombre d'étudiants")
plt.show ()
```

 Exercise 5: Comparing Anxiety Levels Across Different Genders

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Choose your gender', hue='Do you have Anxiety?', palette='Set2')

plt.title("Comparaison des niveaux d'anxiété selon le sexe", fontsize=14)
plt.xlabel("Sexe")
plt.ylabel("Nombre d'étudiants")
plt.show()
```

Exercice 6 : Exploring the Relationship Between Age and Panic Attacks

```python
df['Panic_numeric'] = df['Do you have Panic attack?'].map({'Yes': 1, 'No': 0})
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='Age', y='Panic_numeric', s=100, color='orange', alpha=0.7)
plt.title("Relation entre l'âge et les crises de panique", fontsize=14)
plt.xlabel("Âge")
plt.ylabel("Crises de panique (Oui=1, Non=0)")
plt.yticks([0,1], ['Non','Oui'])
plt.show()
```
