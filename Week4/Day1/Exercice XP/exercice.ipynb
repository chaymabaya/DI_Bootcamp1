Exercice 2 : Creating a Line Plot for Temperature Variation

```python
import matplotlib.pyplot as plt
```

```python
Days = ["lundi " , "Mardi" , "Mercerdi" , "Jeudi" , "Vendredi", "Samedi" , "Dimanche"]
Températeur = [72, 74, 76, 80, 82, 78, 75]
```

```python
plt.figure(figsize = (10,4))
plt.plot( Days , Températeur , marker = "*" , linestyle = '--' , color = "orange" )
plt.grid(True , color = "red" , linestyle = "--" , alpha = 0.2)
plt.xlabel("Days")
plt.ylabel("Températeur")
plt.xticks(Days , Températeur)
plt.legend()
plt.show()
```

Exerice 3 : Visualizing Monthly Sales with a Bar Chart

```python
Month = ["Janvier", "Février", "Mars", "Avril", "Mai", "Juin"]
Sales_Amount = [5000, 5500, 6200, 7000, 7500 , 9000]
plt.bar(Month , Sales_Amount , color= "yellow"  ) 
plt.xlabel("Month")
plt.ylabel("Sales_Amount")
plt.grid(True , linestyle = "--" , alpha = 0.5)
plt.show()
```

Exercice 4 : Data Visualisation

```python
import pandas as pd
```

```python
df = pd.read_csv('sales_data.csv')
```

```python
df
```

```python
total_quantity = df['quantity'].sum
```

```python
total_quantity
```

```python
revenu_by_category = df.groupby('category')['revenue'].sum()
```

```python
best_category = revenu_by_category.idxmax
```

```python
best_category_revenue = revenu_by_category.max()
```

```python
print("Category with highest revenue:", best_category)
print("Revenue generated:", best_category_revenue)
```

```python
df["revenue"].mean()
```

```python
import matplotlib.pyplot as plt
```

```python
df["date"] = pd.to_datetime(df['date'])
```

```python
df["Quarter"] = df["date"].dt.quarter
```

```python
df
```

```python
revenue_by_quarter = df.groupby("Quarter")["revenue"].sum()
```

```python
plt.bar(["Q1", "Q2", "Q3", "Q4"], revenue_by_quarter)
plt.xlabel("Quarters")
plt.ylabel("Total Revenue")
plt.title("Revenue per Quarter")
plt.show()
```

Exercise 5 : Data Visualisation Using MatPlotLib

```python
import numpy as np
x = np.arange(-10 , 10)
y = x**2
plt.plot(x , y , linestyle = '--' , marker = '*' )
plt.xlabel('axe x ')
plt.ylabel("axe y")
plt.show()
```

```python
Produit = ["A" , "B" , "C" , "D"]
ventes = [12, 15 , 45 ,36]
plt.bar(Produit ,  ventes  , color = "pink" , alpha = 1 )
plt.title ("Ventes hebdomadaires des produits A, B, C et D") 
plt.xlabel("Produit")
plt.ylabel("vents")
plt.show()
```

```python
Fruits= ("pommes" , "bananes" , 'cerises' , "dattes")
myexplode = [0, 0, 0.2, 0]
parts = [40, 30, 20, 10]
colors = ["red", "gray", "pink", "brown"]
plt.pie(parts, labels=Fruits, colors=colors, autopct="%1.1f%%", startangle = 90 , explode= myexplode )
plt.legend(loc = "lower left" )
plt.show()

```
